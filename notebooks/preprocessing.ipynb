# Data Preprocessing & NLP Pipeline

This notebook handles text cleaning, tokenization, stop-word removal, lemmatization, and skill taxonomy building.


```python
import os
import re
import json
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

df = pd.read_csv('../data/resume_dataset.csv')
print(f'Loaded dataset with {len(df)} records.')
df.head(3)
```

```python
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = re.sub(r'<[^>]+>', ' ', text)
    text = re.sub(r'http\S+|www\.\S+', ' ', text)
    text = re.sub(r'\S+@\S+', ' ', text)
    text = re.sub(r'[^a-zA-Z0-9\s+#]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['Cleaned_Text'] = df['Resume_str'].apply(clean_text)
print('Text cleaning completed.')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    tokens = text.lower().split()
    filtered = [lemmatizer.lemmatize(w) for w in tokens if w not in stop_words and len(w) > 1]
    return " ".join(filtered)

df['Processed_Text'] = df['Cleaned_Text'].apply(preprocess_text)
df.to_csv('../data/resume_dataset.csv', index=False)
print('Preprocessing pipeline completed & saved to data/resume_dataset.csv')
```
